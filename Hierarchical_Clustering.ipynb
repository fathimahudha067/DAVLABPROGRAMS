```python
# Hierarchical Clustering
#
# Objective:
# To perform Hierarchical Clustering on the Iris dataset
# and visualize the clusters using a dendrogram.
```

```python
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris

from sklearn.preprocessing import StandardScaler

from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
iris = load_iris()

X = iris.data[:, :2]

print("Dataset shape:", X.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Data standardized successfully!")
```

```python
linked = linkage(X_scaled, method='ward')
```

```python
plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")

plt.xlabel("Data Points")

plt.ylabel("Euclidean Distance")

plt.show()
```

```python
model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

clusters = model.fit_predict(X_scaled)

print("Clustering completed successfully!")
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=50
)

plt.title("Hierarchical Clustering (K = 3)")

plt.xlabel("Standardized Sepal Length")

plt.ylabel("Standardized Sepal Width")

plt.show()
```

```python
print("Cluster assignments:")

print(clusters)

print("\nFirst 10 cluster assignments:")

print(clusters[:10])
```

```python
# Hierarchical Clustering:
#
# Hierarchical clustering is an unsupervised machine learning
# algorithm used to group similar data points.
#
# Agglomerative clustering is a bottom-up approach.
# Initially, each data point forms its own cluster.
# The closest clusters are repeatedly merged until
# the required number of clusters is obtained.
#
# A dendrogram is used to visualize the merging process.
#
# Ward's method minimizes the variance within clusters.
```

```python
print("Hierarchical Clustering performed successfully!")

print("Number of clusters:", len(set(clusters)))

print("Total data points:", len(X))
```
