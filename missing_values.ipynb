```python
import numpy as np
import pandas as pd
```

```python
data={
    'Name':['Anu','Ravi','Priya','Kiran','Meena','Arjun'],
    'Age':[21,22,None,23,21,None],
    'Salary':[25000,None,30000,28000,None,32000],
    'Department':['CSE','ECE','CSE',None,'ISE','ECE'],
    'City':['MYS','BNG','MYS','HUBLI','BNG','MYS']
}
```

```python
print(data)
```

```python
df=pd.DataFrame(data)
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].median())
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())

```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
print(data)
```

```python
print("Mean Age:",df['Age'].mean())
```

```python
df=pd.DataFrame(data)
print("before forward fill:")
print(df)
```

```python
df['Department']=df['Department'].ffill()
print("\n after forward fill:")
print(df)
```

```python
encoded_df=pd.get_dummies(df,columns=['Department'],dtype=int)
print(encoded_df)
```
