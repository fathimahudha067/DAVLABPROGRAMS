```python
# K-Means Clustering
#
# Objective:
# To perform K-Means Clustering on the Iris dataset
# and visualize the clusters using the Elbow Method.
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = iris.data[:, :2]

print("Dataset shape:", X.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
inertia = []

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

print("Inertia values:", inertia)

```

```python
plt.plot(range(1, 11), inertia, marker='o')

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

clusters = kmeans.fit_predict(X_scaled)

print("Clustering completed successfully!")
```

```python
plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=clusters,
    cmap='viridis',
    s=50
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200,
    color='red',
    label='Centroids'
)

plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")

plt.title("K-Means Clustering (K = 3)")

plt.legend()

plt.show()
```

```python
# How does the Elbow Method work?

# 1. X-axis represents the number of clusters (K).
# 2. Y-axis represents inertia.
# 3. As K increases, inertia generally decreases.
# 4. Choose a value where the decrease starts slowing down.
# 5. This point is called the Elbow Point.

# In this program, K = 3 is chosen for demonstration.
```

```python
# Example of Inertia Calculation

# Cluster 1
# Study hours: 2, 3, 4
# Centroid = (2 + 3 + 4) / 3 = 3

# Squared distances:
# (2 - 3)^2 = 1
# (3 - 3)^2 = 0
# (4 - 3)^2 = 1

# Cluster 1 inertia = 1 + 0 + 1 = 2


# Cluster 2
# Study hours: 7, 8, 9
# Centroid = (7 + 8 + 9) / 3 = 8

# Squared distances:
# (7 - 8)^2 = 1
# (8 - 8)^2 = 0
# (9 - 8)^2 = 1

# Cluster 2 inertia = 1 + 0 + 1 = 2

# Total inertia = 2 + 2 = 4

print("Total inertia in the example =", 4)
```
